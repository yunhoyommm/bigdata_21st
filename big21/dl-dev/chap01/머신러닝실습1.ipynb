```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
# from sklearn.metrics import *   # 이렇게(*) 하면 안 됨.
# 회귀 모델의 평가 지표 import
from sklearn.metrics import root_mean_squared_error         # RMSE
from sklearn.metrics import r2_score    # r^2   # 회귀 모델의 대표적 성능 평가 지표
from sklearn.metrics import mean_absolute_error             # MAE
from sklearn.metrics import mean_absolute_percentage_error  # MAPE

# 정규화, 표준화
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
# 1. 비즈니스 이해  2. 데이터 이해  3. 데이터 전처리  4. 모델링  5. 평가  6. 배포
# 3. 데이터 전처리
# 3.1 데이터 준비: https://bit.ly/advertising.csv
path = "https://bit.ly/advertising_csv"
data1 = pd.read_csv(path)   # csv 파일 읽어서 데이터프레임 생성
data1.head()
```

```python
# 3.2 데이터 분리: 데이터와 레이블 분리
target = 'Sales'
# 데이터 추출
X = data1.drop(target, axis=1)  # Sales column 삭제 후 데이터 대입
# 레이블 추출
y = data1.loc[:,target]
```

```python
# 3.3 데이터 분할: 학습, 테스트 데이터 분할
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3, random_state=20)
```

```python
# 3.4 표준화 : 0 ~ 1 값으로 변환, MinMaxScaler

# 학습 데이터의 컬럼별 min, max 값 구한다.
scaler = MinMaxScaler()

# # fit(): 식에 필요한 값 구하기
# scaler.fit(X_train)         # (x - min) / (max - min)   x: 데이터 1개   # 컬럼별 minmax 생김(TV, Radio, Newspaper)

# # 변환  (x - min) / (max - min)
# X_train_scaled = scaler.transform(X_train)

# 학습데이터에 대한 변환
X_train_scaled = scaler.fit_transform(X_train)

# 테스트 데이터에 대한 변환(test 데이터 fit 하지 말기. 분포가 달라짐. 기존 fit대로 하기.)
X_test_scaled = scaler.transform(X_test)
```

```python
# 4. 모델링 : 학습(fit()) -> 예측값(predict()) -> 평가(eval)
# 4.1 모델 선택 : LinearRegression(단순선형회귀)
from sklearn.linear_model import LinearRegression
model = LinearRegression()  # 아무 것도 없다. 기본 값으로 처리하겠다.

# 4.2 학습
model.fit(X_train_scaled, y_train)

# 4.3 예측값 추출
preds = model.predict(X_test_scaled)   # 테스트 데이터(new data)에 대한 예측값
```

```python
# 평가
print(f"RMSE : {root_mean_squared_error(y_test, preds)}")
print(f"R2 : {r2_score(y_test, preds)}")
print(f"MAE : {mean_absolute_error(y_test, preds)}")
```
