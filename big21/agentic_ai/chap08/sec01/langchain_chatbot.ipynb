```python
# langchain 사용해서 한 번 대화하는 모델 만들기
from langchain_openai import ChatOpenAI

```

```python
# 1. model 생성
# api_key 생략

model=ChatOpenAI(model="gpt-4o-mini")
```

```python
# # 1. 모델 생성
# # api_key 생략
# model=ChatOpenAI(
#     model="gpt-4o-mini",
#     # temperature=None,  # 0~2
#     # max_tokens=답의 최대 길이(토큰),
#     # timeout=값,  # 기다리는 시간 지정. default=600
#     # max_retries=값,   # 최대 재시작. default=2
#     # api_key,  # 없으면 .env 파일 내 OPENAI_API_KEY 찾아 넣어줌(langchain이)
# )
```

```python
# 2. 질문 -> LLM -> 답변 출력
from langchain_core.messages import HumanMessage    # 사용자의 질문 저장
answer = HumanMessage(content="안녕 나는 윤호야.")      # 질문 생성.
model.invoke([answer])  # 질문 보내고 답변 받기. => AIMessage(content에 답변.)
# model 기능
    # 1) invoke()
    # 2) stream()
    # 3) batch()
    # 4) bind_tools()
```
