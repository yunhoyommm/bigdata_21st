```python
# 화자분리된 텍스트 dataframe으로 loading
import pandas as pd

meeting_note_csv_path="../sec03/싼기타_비싼기타_final.csv"
df_rttm = pd.read_csv(meeting_note_csv_path, sep="|")

df_rttm
```

```python
# 이름 넣기
name_dict = {
    "SPEAKER_00":"AI",
    "SPEAKER_01":"YUN"
}

# df_rttm["name"] = df_rttm["speaker_id"].apply(처리함수)
df_rttm["name"] = df_rttm["speaker_id"].apply(lambda x: name_dict[x])

df_rttm
```

```python
df_rttm.head(3)
```

```python
meeting_note_txt = df_rttm[["start","end","text","name"]].to_json(orient="records", force_ascii=False)
meeting_note_txt
```

```python
# system prompt
system_prompt = f'''
    너는 회의 내용을 요약하는 봇이다.
    아래 회의록을 읽고, 주요 내용을 요약하라.
    결과는 마크다운 형식으로 작성한다.

    아래 형식에 맞추어 작성하라.
    # 회의 제목

    ## 주요 내용

    ## 참석자별 입장

    ## 결정 사항

    ================= 이하 회의록 =================
    {meeting_note_txt}
'''
```

```python
# OpenAI CHAT API
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()
api_key = os.getenv("OPENAPI_AI_KEY")

client = OpenAI(api_key=api_key)    # client => llm => response
```

```python
response=client.chat.completions.create(
    model="gpt-4o",
    temperature=0.1,
    messages=[
        {"role":"system", "content":system_prompt}
    ]
)
response
```

```python
print(response.choices[0].message.content)
```

```python
summary=response.choices[0].message.content
summary=summary.strip() # 양쪽 공백 제거
# 파일에 저장
with open("../audio/guitar_summary2.md", "w", encoding="utf-8") as f:
    f.write(summary)
```

```python
# 화자분리 텍스트 -> LLM(교정) -> 최종 회의록 생성(ms word)
df_rttm.head()
```

```python
df_meeting_note = df_rttm[["start","end","text","name"]].copy()
# 데이터 전처리 => 교정
# NaN 제거
df_meeting_note.dropna(inplace=True)    # inplace=True: 원본에 반영
# df_meeting_note = df_meeting_note.dropna(inplace=False) False: 삭제한 결과 df를 반환
```

```python
# 데이터프레임 => 딕셔너리
meeting_note_dict = df_meeting_note.to_dict(orient="records")   # orient="records": 각 행을 독립된 dictionary로 생성 => 리스트로 묶는다.
meeting_note_dict
```

```python
for row in meeting_note_dict:   # 한 행씩 처리
    # 한 개의 dictionary 안에 있는 text 가져와서 type이 str인지 확인(LLM은 text만 받음.)
    if not isinstance(row["text"], str):
        continue
    # 교정하는 코드: 텍스트(row) => LLM => 교정(correction)
    # 1. correction_system_prompt: 페르소나 + 명령 + text(한 행) + summary + 녹취 전문
    correction_system_prompt = f'''
        rules = 너는 STT 녹취록 교정기다. 규칙:
        1) 잘못 받아적은 단어만 고친다. 말투·어순·내용은 바꾸지 않는다.
        2) 확신이 없으면 원문을 그대로 둔다.
        3) 입력과 같은 id를 가진 JSON만 출력한다: {"items":[{"id":0,"corrected":"..."}, ...]}

        ---------------------------------------------
        회의 요약문 : {summary}
        ---------------------------------------------
        회의 녹취록 전문: {df_meeting_note.to_json(orient="records",force_ascii=False)}
        ---------------------------------------------
        확인할 메시지 원본: {row["text"]}
    '''
    # end prompt
    # 2. response
    response = client.chat.completions.create(
        model="gpt-6-astra",     # token limit : 분당 30000
        # temperature=0.1,    # 
        messages=[
            {"role":"system","content":correction_system_prompt}
        ]
    )

    correction = response.choices[0].message.content    # 한 개 행에 대한 교정 결과
    if ":" in correction:   # "수정 ':' 내용"과 같은 형식으로 반환되는 경우 해결? 너무 지엽적인 느낌....
        correction = correction.split(":")[-1]

    correction = correction.strip()

    print(row["name"])  # 누가 말했고 출력
    print(row["text"])  # 교정 전 원본
    print(correction)   # 교정된 대화
    print("-"*100)

    row["corrected_text"] = correction  # row: 딕셔너리, 'corrected_text' 키에 값은 correction
```
