```python
import base64

# 1. 이미지를 인코딩하는 함수
def encode_image(image_path):
    with open(image_path, "rb") as image_file:
        # image_encode = image_file.read()


        return base64.b64encode(image_file.read()).decode("utf-8")
```

```python
image_path="../data/images/local_stitch_terrarosa.jpg"

# 이미지를 base64로 인코딩
base64_image = encode_image(image_path=image_path)

print(base64_image)
```

```python
# data:image/jpeg;base64
# data : LLM에게 보낼 데이터 형식
# data:image/jpeg : jpeg 형식의 이미지를 보낼 거야.
# base64 : 인코딩 형식 전달해 줘야.
```

다만 이를 곧바로 “한국의 연구개발 경쟁력이 약해졌다”라고 해석하기는 어렵습니다.

- **GDP 대비 투자 비중도 상승**했습니다.
- 달러 기준 총액은 **환율의 영향을 크게 받습니다.**
- 국가별 자료의 기준연도가 다르고, 통계 개정이나 집계 방식 변화도 순위에 영향을 줄 수 있습니다.

## 4. 비교할 때 주의할 점

한국의 수치는 각각 **2021년과 2022년**이지만, 다른 나라까지 모두 같은 연도를 비교한 것은 아닙니다. 예를 들어 두 번째 그래프의 미국·중국·일본·독일은 2021년, 호주는 2019년, 러시아는 2020년 자료입니다.

따라서 두 그래프는 **각 시점에 확보된 국가별 최신 통계의 비교**이며, 모든 국가의 2021→2022년 변화를 동일 조건에서 보여주는 자료는 아닙니다.

**정리하면, 한국은 2022년에 연구개발 투자 규모를 원화 기준으로 확대했고, 경제 규모 대비 투자 수준도 높였습니다. 달러 기준 금액 감소와 총액 순위 하락은 이러한 투자 확대와 구분해서 봐야 합니다.**

```python
def image_quiz(image_path):
    base64_image = encode_image(image_path) # 이미지를 base64로 인코딩


    quiz_prompt = """
    제공된 이미지를 바탕으로, 다음과 같은 양식으로 퀴즈를 만들어주세요.
    정답은 1~4 중 하나만 해당하도록 출제하세요.
    아래는 예시입니다.
    ----- 예시 -----


    Q: 다음 이미지에 대한 설명 중 옳지 않은 것은 무엇인가요?
    - (1) 베이커리에서 사람들이 빵을 사고 있는 모습이 담겨 있습니다.
    - (2) 맨 앞에 서 있는 사람은 빨간색 셔츠를 입고 있습니다.
    - (3) 기차를 타기 위해 줄을 서 있는 사람들이 있습니다.
    - (4) 점원은 노란색 티셔츠를 입고 있습니다.
       
    정답: (4) 점원은 노란색 티셔츠가 아닌 파란색 티셔츠를 입고 있습니다.
    (주의: 정답은 1~4 중 하나만 선택되도록 출제하세요.)
    ======
    """


    messages = [
        {
            "role": "user",
            "content": [
                {"type": "text", "text": quiz_prompt},
                {
                    "type": "image_url",
                    "image_url": {
                        "url": f"data:image/jpeg;base64,{base64_image}",
                    },
                },
            ],
        }
    ]


    response = client.chat.completions.create(
        model="gpt-4o",  # 응답 생성에 사용할 모델을 지정
        messages=messages # 대화 기록을 입력으로 전달
    )


    return response.choices[0].message.content  # 해당 이미지에 대한 문제 리턴
```

```python
q=image_quiz("../data/images/local_stitch.jpg")
print(q)
```

```python
from glob import glob
# glob function(디렉토리) : 해당 디렉토리 안에 있는 파일 목록 리스트로 반환
glob("../data/images/*.jpg")
```

```python
# 8개 이미지에 문제 8개 생성



txt=''  # 문제들을 계속 붙여 나가기 위해 빈 문자열로 담는 변수 선언
no=1    # 문제 번호를 위해 선언

for g in glob("../data/images/*.jpg"):  # g는 파일명에 대한 문자열.
    try:    # 파일이 없을 수도 있으니까..
        q=image_quiz(g)     # LLM에 문제 생성 요청. LLM은 Error 생길 수 있으니 try 해야 함.
    except Exception as e:
        print(e)    # 원인 출력
        continue    # 이 줄이 실행된다는 것은 에러 발생했다는 의미. 한 장의 이미지를 LLM이 처리 못 했다.
    # 이 아래 코드는 정상적으로 문제 생성되었을 때 실행.
    # 1. 문제 번호
    divider = f"## 문제 {no}\n\n"
    print(divider)

    # 2. 문제 번호를 txt(전체 문제를 저장하는 변수)에 추가
    txt += divider

    # 3. 해당 문제의 이미지 파일 경로를 저장
    filename = os.path.basename(g)  # 이미지 파일명 추출(파일명+확장자)
    txt += f"![image]({filename})\n\n"

    # 4. 문제 저장
    print(txt)
    print(q)
    txt += q + '\n\n--------------------------\n\n'

    no += 1     # 문제 번호 증가
```
