```python
# pip install tavily-python : 파이썬에서 타빌리 사용할 수 있게 한 라이브러리
# 타빌리 : 검색 + LLM 기능
from langchain_community.tools import TavilySearchResults       # 사실 tavily_python 쓴 것.

# 검색 오브젝트 생성
tavily_search = TavilySearchResults(
    max_results=5,              # 최대 검색 결과 개수
    # search_depth='advanced'    
)

# 질문-> 검색
response = tavily_search.invoke(
    {"query":"2027년 한국 "}
)

for res in response:
    print(res)
    
```

```python
# 주피터 노트북 전용 명령(보기 좋게 출력)
from IPython.display import JSON
JSON(response).data
```

```python
# tavily 직접 사용
from tavily import TavilyClient # 요청하는 주체 => http(network) => Tavily Server => 검색 결과 => client

# Tavily client 생성 => 서버에 요청하는 주체
client = TavilyClient()

# 검색 요청
content = client.search(
    query="2027년 한국 경제 전망",
    search_depth="advanced",
    include_raw_content=True
)

JSON(content).data
```

```python
# from langchain_community.adapters.openai import convert_openai_messages
from langchain_openai import ChatOpenAI


llm = ChatOpenAI(model="gpt-4o")


# Step 1. 질문에 대한 자료를 TavilyClient로 가져오기
query = "현대자동차 미국 시장 2027년 전망"


# client => tavily
content = client.search(
    query,
    include_raw_content=False,
    search_depth="advanced",
    max_results=2
)["results"]


content


```

```python
# Step 2. 프롬프트 설정
prompt = [{
    "role": "system",
    "content":  (
        "당신은 신문기사를 쓰는 기자 AI입니다.  \n"
        "당신은 주어진 정보를 바탕으로 객관적이고 체계적으로 작성된 기사를 써야 합니다. \n"
    )
}, {
    "role": "user",
    "content": (
        f'정보: """{content}"""\n\n'
        f'위의 정보를 사용하여, 다음 질문에 대해 자세한 보고서를 한국어로 작성하세요: "{query}" \n'
        '—신문기사 형식을 사용하되, MLA를 준수하는 markdown 문법을 사용해주세요.'
        '—활용한 자료는 출처를 명시하세요.'
    )
   
}]
```

```python
# Step 3. 오픈AI를 랭체인으로 실행하고 출력
report = llm.invoke(prompt).content
print(report)
```
