```python
# 유튜브에서 미국 대선 관련 검색
from youtube_search import YoutubeSearch

videos = YoutubeSearch(
    "chatgpt 오픈소스",
    max_results=5
).to_dict()

videos
```

```python
video_url = "https://youtube.com" + videos[-1]["url_suffix"]
video_url
```

```python
from langchain_community.document_loaders import YoutubeLoader


loader = YoutubeLoader.from_youtube_url(
    video_url,
    language=['ko', 'en'] # 자막 언어
)


loader.load()
```

```python
import time


for v in videos:    # 리스트에서 딕셔너리를 하나씩 꺼내서 v에게 준다.
    # url_suffix를 이용하여 video_url을 만듭니다.
    v['video_url'] = 'https://youtube.com' + v['url_suffix']
   
    # YoutubeLoader를 이용하여 비디오를 로드합니다.
    loader = YoutubeLoader.from_youtube_url(
        v['video_url'],
        language=['ko', 'en']
    )


    v['content'] = loader.load()    # 자막을 추출해서 content에 넣겠다.
    time.sleep(2)   # 2초 간 정지. client => server 요청 => 자막 => client
videos
```

```python
print('총 영상 수:', len(videos))
# 영상 길이가 60분 이하인 영상만 남깁니다.
videos = [v for v in videos if len(v['duration'].split(':')) < 3]
print('60분 이하 영상 수:', len(videos))
```

```python
# 자막 내용 요약 -> LLM
# 랭체인, openai 임포트
from langchain_openai import ChatOpenAI


model = ChatOpenAI(model="gpt-4o-mini")
model.invoke('안녕?') # 언어 모델이 잘 설정되었는지 테스트
```

```python
# 자막 => 질문에 넣어야 함. => 템플릿
from langchain_core.prompts import ChatPromptTemplate
# from langchain.chains.combine_documents import create_stuff_documents_chain


# 동영상 요약 프롬프트 작성
prompt = ChatPromptTemplate.from_messages(
    [("system", "다음 영상에 대한 요약을 한국어로 만들어줘 :\\n\\n{context}")]
)
# chain = create_stuff_documents_chain(model, prompt)
chain = prompt | model


result = chain.invoke({"context": videos[0]['content']})
result
```

```python
# 5개 비디오 요약 처리
# 모든 비디오에 대해 요약을 생성
from tqdm import tqdm # tqdm은 진행 상황을 보여주는 라이브러리


for v in tqdm(videos):
    v['summary'] = chain.invoke({"context": v['content']})
   
videos
```
